# 1. Applying BoxCox Transformation

```python
import numpy
```

```python
data_pd = data_pd.drop(columns = "Datetime")
```

```python
data_pd.head()
```

```python
skewness = data_pd.skew()
```

```python
skewness
```

```python
from scipy.stats import boxcox
```

```python
values = []
for i,v in enumerate(skewness):
    if -1<v<1:
        continue
    else:
        values.append(skewness.index[i])
```

```python
values
```

```python
lambdas = {}
logged =[]
for i in values:
    try:
        boxcox_ = boxcox(data_pd[i])
        lambdas[i] = boxcox_[1]
        data_pd[i] = boxcox_[0]
    except Exception as e:
        print(f"The column {i} has negative values!! So performing log1p transform")
        data_pd[i] = np.log1p(data_pd[i])
        logged.append(i)
```

```python
data_pd.head()
```

```python
lambdas
```

```python
data_pd.skew()
```

```python
X_bc_data = data_pd.drop(columns=["C6H6(GT)"],axis=1)
```

```python
y_bc_data = data_pd["C6H6(GT)"]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.metrics import r2_score,mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
X_bc_ss = StandardScaler()
```

```python
xtrain_bc,xtest_bc,ytrain_bc,ytest_bc = train_test_split(X_bc_data,y_bc_data,test_size=0.3)
```

```python
xtrain_bc_scaled = X_bc_ss.fit_transform(xtrain_bc)
xtest_bc_scaled = X_bc_ss.transform(xtest_bc)
```

```python
lr_bc = LinearRegression()
```

```python
lr_bc.fit(xtrain_bc_scaled,ytrain_bc)
```

```python
ypred_bc = lr_bc.predict(xtest_bc_scaled)
```

```python
r2_score(ytest_bc,ypred_bc)
```

```python
mean_squared_error(ytest_bc,ypred_bc)
```

# Lasso Regression

```python
xtrain_bc,xtest_bc,ytrain_bc,ytest_bc = train_test_split(X_bc_data,y_bc_data,test_size=0.3)
```

```python
xtrain_bc.head()
```

```python
from sklearn.model_selection import KFold
```

```python
kf = KFold(n_splits=5,shuffle=True,random_state = 42)
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
```

```python
from sklearn.linear_model import Lasso
```

```python
lasso_pipeline = Pipeline([
    ('poly',PolynomialFeatures()),
    ('scaler',StandardScaler()),
    ('lasso',Lasso())
])
```

```python
param_grid_lasso = {
    'poly__degree' : [2,3,5],
    'lasso__alpha' : [0.1,0.001,1,10]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_lasso = GridSearchCV(lasso_pipeline, param_grid_lasso,cv=kf,scoring='neg_mean_squared_error')
```

```python
grid_lasso.fit(xtrain_bc,ytrain_bc)
```

```python
best_lasso = grid_lasso.best_estimator_
```

```python
grid_lasso.best_params_
```

```python
y_pred_lasso = best_lasso.predict(xtest_bc)
```

```python
from sklearn.metrics import r2_score, mean_squared_error
```

```python
mse_lasso = mean_squared_error(ytest_bc,y_pred_lasso)
```

```python
r2_lasso = r2_score(ytest_bc,y_pred_lasso)
```

```python
print(f"MSE Lasso Regression = {mse_lasso}")
print(f"R2_Score Lasso Regression = {r2_lasso}")
```
